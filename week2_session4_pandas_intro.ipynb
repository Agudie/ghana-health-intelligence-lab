```python
import pandas as pd

data = {
    "facility": ["Suhum Government Hospital", "Koforidua Polyclinic",
                 "Nsawam Health Centre", "Asamankese District Hospital"],
    "malaria": [45, 85, 60, 30],
    "total": [120, 250, 150, 200]
}

df = pd.DataFrame(data)
df
```

```python
df["malaria_pct"] = (df["malaria"] / df["total"] * 100).round(1)
df
```

```python
df[df["malaria_pct"] > 30]
df.sort_values("malaria_pct", ascending=False)
```

```python
df["burden"] = df["malaria_pct"].apply(
    lambda x: "high" if x > 30 else "low"
)
```

```python
high_burden = df[df["burden"] == "High"]
high_burden = high_burden.sort_values(
    "malaria_pct",
    ascending=False
)
high_burden
```
