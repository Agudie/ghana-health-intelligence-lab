## Week 1: Setup and First Steps

I am learning Python to apply data science and artificial intelligence to public health challenges in Ghana and Africa.

```python
# Week 1 Session 3 - Lists, Dictionaries, Loops

facilities = {
    "Suhum Government Hospital": {"malaria": 45, "total": 120},
    "Koforidua Polyclinic": {"malaria": 85, "total": 250},
    "Nsawam Health Centre": {"malaria": 60, "total": 150},
    "Asamankese District Hospital": {"malaria": 30, "total": 200}
}

for name, data in facilities.items():
    malaria_cases = data["malaria"]
    total = data["total"]
    proportion = (malaria_cases / total) * 100

    print(name, "_", round(proportion, 1), "% malaria")

    if proportion > 30:
      print(" -> High burden")
    else:
      print(" -> Moderate / Low burden")
```
